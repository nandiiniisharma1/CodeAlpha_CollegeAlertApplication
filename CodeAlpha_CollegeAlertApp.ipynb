```python

import sqlite3, threading, time, subprocess, re
from datetime import datetime
from flask import Flask, request, jsonify, send_file
from IPython.display import display, HTML

app = Flask(__name__)
DB = 'college_alerts.db'

def init_db():
    con = sqlite3.connect(DB)
    con.execute("""
        CREATE TABLE IF NOT EXISTS alerts (
            id          INTEGER PRIMARY KEY AUTOINCREMENT,
            title       TEXT NOT NULL,
            description TEXT NOT NULL,
            type        TEXT NOT NULL,
            date        TEXT NOT NULL,
            time        TEXT NOT NULL,
            venue       TEXT NOT NULL,
            created     TEXT NOT NULL
        )
    """)
    if con.execute('SELECT COUNT(*) FROM alerts').fetchone()[0] == 0:
        samples = [
            ('Mathematics Mid-Term Exam',
             'Unit 1-4 syllabus. No open-book. Bring calculator and student ID.',
             'Exam', '2025-07-15', '10:00 AM', 'Exam Hall B', datetime.now().isoformat()),
            ('AI & Machine Learning Seminar',
             'Guest lecture by Dr. Sharma from IIT Delhi. Q&A session included.',
             'Seminar', '2025-07-18', '02:00 PM', 'Auditorium Block C', datetime.now().isoformat()),
            ('Annual College Fest — TechFusion 2025',
             'Coding, robotics and cultural events. Prizes worth Rs.50,000!',
             'Fest', '2025-07-25', '09:00 AM', 'Main Ground', datetime.now().isoformat()),
            ('Holiday Notice — Independence Day',
             'College closed 15th August. Classes resume 16th August 2025.',
             'Notice', '2025-08-15', 'All Day', 'N/A', datetime.now().isoformat()),
        ]
        con.executemany(
            'INSERT INTO alerts (title,description,type,date,time,venue,created) VALUES (?,?,?,?,?,?,?)',
            samples
        )
    con.commit()
    con.close()

init_db()

@app.route('/')
def index():
    return send_file('template.html')

@app.route('/api/alerts', methods=['GET'])
def get_alerts():
    con = sqlite3.connect(DB)
    rows = con.execute('SELECT * FROM alerts ORDER BY id DESC').fetchall()
    con.close()
    return jsonify([
        {'id':r[0],'title':r[1],'description':r[2],'type':r[3],'date':r[4],'time':r[5],'venue':r[6]}
        for r in rows
    ])

@app.route('/api/alerts', methods=['POST'])
def add_alert():
    d = request.json
    con = sqlite3.connect(DB)
    con.execute(
        'INSERT INTO alerts (title,description,type,date,time,venue,created) VALUES (?,?,?,?,?,?,?)',
        (d['title'], d['description'], d['type'], d['date'], d['time'], d['venue'], datetime.now().isoformat())
    )
    con.commit()
    con.close()
    return jsonify({'status': 'ok'}), 201

@app.route('/api/alerts/<int:aid>', methods=['DELETE'])
def delete_alert(aid):
    con = sqlite3.connect(DB)
    con.execute('DELETE FROM alerts WHERE id=?', (aid,))
    con.commit()
    con.close()
    return jsonify({'status': 'deleted'})


def run_flask():
    app.run(port=5000, use_reloader=False, debug=False)

t = threading.Thread(target=run_flask, daemon=True)
t.start()
time.sleep(2)
print("Flask is running on port 5000...")


!npm install -g localtunnel --silent 2>/dev/null
import subprocess, threading

tunnel_url = [None]

def start_tunnel():
    proc = subprocess.Popen(
        ['lt', '--port', '5000'],
        stdout=subprocess.PIPE, stderr=subprocess.PIPE, text=True
    )
    for line in proc.stdout:
        line = line.strip()
        if 'loca.lt' in line or 'localtunnel' in line.lower():
            import re
            match = re.search(r'https?://[\w.-]+\.loca\.lt', line)
            if match:
                tunnel_url[0] = match.group()
                break

tunnel_thread = threading.Thread(target=start_tunnel, daemon=True)
tunnel_thread.start()


for _ in range(20):
    time.sleep(1)
    if tunnel_url[0]:
        break

if tunnel_url[0]:
    url = tunnel_url[0]
    print()
    print("=" * 58)
    print("  COLLEGE ALERT APP IS LIVE!")
    print("=" * 58)
    print(f"  Open in browser: {url}")
    print()
    print("  NOTE: LocalTunnel may show a warning page first.")
    print("  Just click 'Click to Continue' to open the app.")
    print()
    print("  Record this for your CodeAlpha LinkedIn post!")
    print("=" * 58)
    display(HTML(f'''<div style="margin:16px 0;padding:16px;background:#E3F2FD;border-radius:8px;font-family:sans-serif">
    <b style="color:#1565C0">Your app is live!</b><br><br>
    <a href="{url}" target="_blank" style="color:#1565C0;font-size:16px">{url}</a><br><br>
    <small style="color:#666">If you see a warning page, click "Click to Continue"</small>
    </div>'''))
else:

    from google.colab import output
    print()
    print("LocalTunnel timed out. Using Colab port forwarding instead.")
    print("Run the line below to open the app:")
    output.serve_kernel_port_as_window(5000)
```
